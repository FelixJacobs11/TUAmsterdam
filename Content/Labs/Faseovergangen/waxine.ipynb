# Waxinekaarsje

Als je een waxinekaarsje aan steekt en er een glas overheen zet, zal het kaarsje uit gaan. 
Als je dat waxinekaarsje laat drijven op water terwijl je er het glas er overheen zet, zal het water in het glas stijgen.
Op het internet zwerven er verschillende verklaringen voor dit fenomeen rond.
1. Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.
2. Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af waardoor de druk afneemt en er water omhoog wordt gezogen.
3. Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water omhoog gezogen.

```{exercise}
Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.
```

Eerst hebben wij de proef uitgevoerd met een schaaltje water, een waxinelampje en een maatcilinder. We hebben het waxinelichtje aan gedaan en in het water gedaan (Het water kwamt tot net onder de rand van de kaars). Toen hebben we de maatcilinder eroverheen gedaan (omgekeerd) zonder dat die de bodem raakte waardoor er nog wel water ingezogen kon worden. Wij namen zoals verwacht waar dat het water omhoog ging. De volgende conclusies kunnen getrokken worden:
- Er is geen toevoer van zuurstof omdat de ruimte in de maatcilinder afgesloten is door de maatilinder en het water.
- Na een tijdje wordt hij omhooggetrokken, dit begint als de kaars nog brandt
- De kaars gaat uit op een gegeven moment
- Als de kaars uitgaat stijgt de kaars explosief
- Bij herhaalde meting blijft hij ongeveer hetzelfde.
- Het water buiten de maatcilinder toont geen opmerkzaamheden behalve dat het wat omlaag gaat

Daarna hebben wij gekeken naar de impact van meerdere lampjes. Hiervoor hebben we een maatbeker genomen (die wat breder is). We hebben gekeken naar het verschil als je 1, 2 of 3 lampjes erin doet. Voor de rest hebben we dezelfde procedure als hierboven gebruikt. Wij bevonden het volgende:
-Bij meer lichtjes ging het water significant meer omhoog: 1 het laagste, 2 middel, 3 het hoogste
- Verder zijn dezelfde conclusies getrokken als bij de eerdere proef

Wij hebben ook gekeken naar het gebruik van een kleiner maatbeker

Hypothese 1: Kan niet want je ziet dat met drie waxinelichtjes het water hoger komt dan bij een terwijl de hoeveelheid zuurstod hetzelfde is in de maatbeker.
Hypothese 2: Kan niet want stel hij zou door afkoelen omhoog gaan zou hij eerst omlaag moeten zijn gegaan en daarna niet hoger dan aan origineel
Hypothese 3: Klopt, deze theorie wordt bevestigd door dat bij meer lichtjes (en dus meer waterdamp) er meer condens was en hij dus hoger kwam. Hij wordt niet weersproken door ons experiment.
