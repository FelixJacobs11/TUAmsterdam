# Theorie

Volgens [Newton's wet van afkoeling](https://nl.wikipedia.org/wiki/Afkoelingswet_van_Newton) is de snelheid waarmee een voorwerp afkoelt evenredig met het verschil in de temperatuur van het voorwerp ($T$) en de omgeving ($T_0$). We kunnen dit schrijven als:

$$
    \dot{Q} = -hA(T(t) - T_0),
$$

waarin 

- $\dot{Q}$ de warmtestroom in $\mathrm{W}$,
- $A$ het oppervlak waardoor koeling optreedt in $\mathrm{m}^2$,
- $h$ de warmteoverdrachtscoëfficiënt in $\mathrm{W/(m^2 K)}$.

Dit levert de differentiaalvergelijking

$$
    C\dot{T} = -hA(T(t) - T_0),
$$

met $C$ de warmtecapaciteit in $\mathrm{J/kg}$. Herschrijven met $\tau = \frac{C}{hA}$ levert:

$$
    -\tau\dot{T} = T(t) - T_0,
$$ (eq:dv1)

met als oplossing:

$$
    T(t) - T_0 = (T(0) - T_0)\text{e}^{-t/\tau}.
$$ (eq:dv1_opl)

We kunnen hieruit dus concluderen dat $\tau$ de karakteristieke tijdsduur is waarin de temperatuur van de buis een factor $\text{e}$ verlaagd ten opzichte van de omgevingstemperatuur. 

```{exercise} Oplossing DV controleren 
:label: ex-dvopl
Controleer bovenstaande door vergelijking [](#eq:dv1_opl) te substitueren in [](#eq:dv1).
```

````{solution} ex-dvopl
ALs je de functie: $T(t)-T_0 = (T(0)-T_0)\,\mathrm{e}^{-t/\tau}$ differentieert naar de tijd krijg je $\dot T(t) = -\frac{1}{\tau}(T(0)-T_0)\,\mathrm{e}^{-t/\tau}.$

Substitueer het dan in de differentiaalvergelijking:$-\tau \dot T(t) = -\tau \Bigl(-\frac{1}{\tau}(T(0)-T_0)\,\mathrm{e}^{-t/\tau}\Bigr) = (T(0)-T_0)\,\mathrm{e}^{-t/\tau}.$
Uit $(T(0)-T_0)\,\mathrm{e}^{-t/\tau} = T(t)-T_0$
volgt dat de vergelijking klopt

````T

We zijn hier voor het gemak uitgegaan van een $h$ die onafhankelijk is van de temperatuur. We weten echter dat warmtetransport door straling niet lineair gaat, maar als 

$$
    \dot{Q}_s = \epsilon \sigma A (T^4 - T_0^4).
$$ 

Voor kleine temperatuurverschillen ($\Delta T = T - T_0$) is dit te vereenvoudigen tot

$$
    \dot{Q}_s = \epsilon \sigma A ((T_0+\Delta T)^4 - T_0^4) \approx \epsilon \sigma 4A T_0^3 \Delta T.
$$

Zolang $\Delta T$ dus relatief klein is ten opzichte van $T_0$, kunnen we $h$ dus inderdaad als een constante beschouwen. 

```{exercise} Uitwerken
:label: ex_uitw

Laat zien dat bovenstaande geldt door eerst $(T_0+\Delta T)^4$ uit te schrijven, te substitueren en dan te bedenken dat $\Delta T$ klein is en $\Delta T^2$ dus nog kleiner.
```

```{solution} ex_uitw
Het uitwerken van $(T_0 + \Delta T)^4 = T_0^4 + 4 T_0^3 \Delta T + 6 T_0^2 (\Delta T)^2 + 4 T_0 (\Delta T)^3 + (\Delta T)^4.$
Vervolgnes substitutie $\dot Q_s = \epsilon \sigma A \Bigl((T_0+\Delta T)^4 - T_0^4\Bigr) = \epsilon \sigma A \Bigl(4 T_0^3 \Delta T + 6 T_0^2 (\Delta T)^2 + 4 T_0 (\Delta T)^3 + (\Delta T)^4 \Bigr)$
Als T erg klein is kun je de hogere waardes verwaarlozen dit geeft $\dot Q_s \approx \epsilon \sigma A \, 4 T_0^3 \Delta T$
Hieruit zien we dat voor kleine temperatuurverschillen de benadering klopt
```

# Resultaten

`````python
#Importeren libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
`````

`````python
Diameter_binnen= 44.6e-3 #m #Diameter van de binnenkant
u_diameter_binnen=0.05e-3 #m #onzekerheid diameter van binnenkant
Buis=149.4e-3 #m #lengte van de buis
Dikte=1.5e-3 #m #Dikte van de buis (het metalen deel)
u_dikte=0.05e-3 #m #onzekerheid dikte van de buis

buitenoppervlak=((Diameter_binnen+2*Dikte)+(Diameter_binnen))*np.pi*Buis #m2 
u_buitenoppervlak=np.sqrt((u_diameter_binnen/Diameter_binnen)**2+(u_dikte/Dikte)**2)*buitenoppervlak #m2 #onzekerheid
    
C=8.8e2 #J/(kg K) #Warmtecapaciteit
rho=2.7e3 #kg/m^3
m3=(((Diameter_binnen+2*Dikte)**2)-(Diameter_binnen)**2)*np.pi*Buis #m^3    
warmtecapaciteit = (rho*m3)*C #J/K #Warmtecapaciteit voor deze buis

T_omg=23.5 #graden Celsius
`````

`````python
#Resultaten analyseren
#eerste resultaten van verticale buis zonder dop
#Functie definieren
def exp_func(t, A, tau):
    # A is verschiltemperatuur met omgeving aan start
    # tau is de karakteristieke tijd voor de koeling
    # T_omg is de omgevingstemperatuur
    return (A * np.exp(-t/tau) + T_omg)  

#data laden 
data1=np.loadtxt('C:\\TUDelftProgrameren\\Tweede_meting_temperatuur.txt')

#Data sorteren 
verticaal_t=data1[:,0]  #s
verticaal_T=data1[:,1] #C


#Curvefit maken
popt, pcov = curve_fit(exp_func, verticaal_t, verticaal_T, p0=[50, 1000], maxfev=5000)
A_exp, tau_exp = popt #waardes gesorteerd
u_tau=np.sqrt(pcov[1][1]) #onzekerheid berekenen van karakteristieke tijd

#fit data aanmaken
x_fit=np.linspace(min(verticaal_t), max(verticaal_t), 500) #x waardes maken in secondes
y_fit = exp_func(x_fit, *popt) #fit

#plotten van data
plt.figure()
plt.xlabel('$t$ [s]')
plt.ylabel('$T$ [C]')
plt.plot(verticaal_t, verticaal_T, 'bo', label='measurement')
plt.plot(x_fit, y_fit, 'r-', label='$T = %0.2f e^{-t/%0.4f} + %0.2f$' % (A_exp, tau_exp, T_omg))
plt.legend()
plt.savefig(r"C:\TUDelftProgrameren\thermolab\Figures\grafiek_koelbuis.png", dpi=450) #opslaan figuur
plt.show()

#warmteoverdrachtscoefficient berekenen en onzekerheid
h_exp = (warmtecapaciteit) / (tau_exp * buitenoppervlak) #W/m^2 K
u_h_exp=np.sqrt((0.1/warmtecapaciteit)**2+(u_tau/tau_exp)**2+(u_buitenoppervlak/buitenoppervlak)**2)*h_exp #W/m^2 K
                
#Resultaten presenteren
print('De karakteristieke tijd zonder dop is:', tau_exp, '+/-',u_tau,'s')
print('De warmteoverdrachtscoefficient zonder dop is:',h_exp,'+/-',u_h_exp,'W/m^2 K') # warmteoverdrachtscoëfficiënt in W/m^2 K
`````

`````python
# Nu de analyse van met een dop
#Data laden
data2=np.loadtxt('C:\\TUDelftProgrameren\\Thermolab_warmte_buis1.txt',skiprows=1)

#sorteren
dop_t=data2[:,0]
dop_T=data2[:,1]

#fitten
popt, pcov = curve_fit(exp_func, dop_t, dop_T, p0=[50, 1000], maxfev=5000)

#waardes sorteren
A_dop_exp, tau_dop_exp = popt
u_tau_dop=np.sqrt(pcov[1][1]) #onzekerheid berekenen van karakteristieke tijd

#fit waardes maken
xdop_fit=np.linspace(min(dop_t),max(dop_t),500)
ydop_fit = exp_func(xdop_fit, *popt)

#plotten
plt.figure()
plt.xlabel('$t$ [s]')
plt.ylabel('$T$ [C]')
plt.plot(dop_t, dop_T, 'bo', label='measurement')
plt.plot(xdop_fit, ydop_fit, 'r-', label='$T = %0.2f e^{-t/%0.4f} + %0.2f$' % (A_exp, tau_exp, T_omg))
plt.legend()
plt.savefig(r"C:\TUDelftProgrameren\thermolab\Figures\grafiek_koelbuis_dop.png", dpi=450) #grafiek opslaan
plt.show()

#warmteoverdrachtscoefficient berekenen
h_dop_exp = (warmtecapaciteit) / (tau_dop_exp * buitenoppervlak) #W/m^2 K
u_h_dop_exp=np.sqrt((0.1/warmtecapaciteit)**2+(u_tau_dop/tau_dop_exp)**2+(u_buitenoppervlak/buitenoppervlak)**2)*h_exp #W/m^2 K

#presenteren waardes
print('De karakteristieke tijd met een dop is', tau_dop_exp,'+/-',u_tau_dop, 's')
print('De warmteoverdrachtscoefficient zonder dop is:',h_dop_exp,'+/-',u_h_dop_exp,'W/m^2 K') # warmteoverdrachtscoëfficiënt in W/m^2 K
`````

Waarschijnlijk komt het grootste deel van de overdracht door convectie er is namelijk een groot buitenoppervlak met de lucht, de dop zorgt voor minder convectie omdat er dan een kleiner buitenoppervlak is van de buis en omdat de dop ervoor zorgt dat er minder nieuwe lucht in de buis kan komen waardoor de lucht relatief warm blijft. Straling zal er altijd wel zijn maar deze is hier minimaal door lage temperatuur. Geleiding is ook een klein deel door de grijpers en later ook de dop.

# Discussie en conclusie
Uit dit experiment blijkt dat de warmtegeleidings coefficient hoger is bij de buis met één dop dan de buis zonder dop. Zonder dop gaf een coefficient van 13.8 +/- 0.5 W/(m^2 K) en met een dop 17.7 +/- 0.6 W/(m^2 K), deze metingen zien er goed uit en zijn relatief significant. De metingen kwamen ook erg goed overeen met de theorie. Redenen voor de hogere coefficient met dop zou de manier van meten kunnen zijn. Doordat de lucht aan de binnenkant van de de buis warmer is (de warme lucht kan er moeilijker uit door de dop) vindt er minder convectie aan de binnenkant plaats. Dit maakt de buis relatief warmer en dus is er wel mogelijkheid tot meer convectie aan de buitenkant (groter temperatuurverschil). Wij hebben de temperatuur ook steeds aan de buitenkant gemeten (deze daalde dan dus sneller) en niet aan de binnenkant. In de toekomst zou hier naar gekeken kunnen worden door misschien ook te kijken naar de temperatuur aan de binnenkant en het effect van twee doppen (aan beide kanten).
